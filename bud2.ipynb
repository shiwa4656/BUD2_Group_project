```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("norwegian_financials_raw.csv")

df
```

## 2 Ratio Analysis

## 2.1 Profitability Ratios

### Operating Profit Margin



```python
df["operating_profit_margin"] = (
    df["operating_profit_ebit_nok_m"]
    / df["operating_revenue_nok_m"]
) * 100

profit_margin_table = df.loc[
    df["year"].between(2021, 2025),
    [
        "year",
        "operating_revenue_nok_m",
        "operating_profit_ebit_nok_m",
        "operating_profit_margin"
    ]
].copy()

profit_margin_table
```

### Return on Capital Employed (ROCE)

```python
##Calculating Capital Employed
df["capital_employed"] = (
    df["total_equity_nok_m"]
    + df["total_noncurrent_liabilities_nok_m"]
)

df[[
    "year",
    "total_equity_nok_m",
    "total_noncurrent_liabilities_nok_m",
    "capital_employed"
]]

## calculate the average capital employed for each year and the ROCE for each year

df["average_capital_employed"] = (
    df["capital_employed"]
    + df["capital_employed"].shift(1)
) / 2

## ROCE Calculation:
df["roce"] = (
    df["operating_profit_ebit_nok_m"]
    / df["average_capital_employed"]
) * 100

##Displaying the ROCE table for the years 2021-2025

roce_table = df.loc[
    df["year"].between(2021, 2025),
    [
        "year",
        "operating_profit_ebit_nok_m",
        "capital_employed",
        "average_capital_employed",
        "roce"
    ]
].copy()

roce_table
```

```python
## Profitability table (Operating Profit Margin and ROCE) for the years 2021-2025

profitability = df.loc[
    df["year"].between(2021, 2025),
    [
        "year",
        "operating_profit_margin",
        "roce"
    ]
].copy()

profitability = profitability.rename(columns={
    "operating_profit_margin": "Operating Profit Margin (%)",
    "roce": "ROCE (%)"
})

profitability.round(2)
```
