```python
import pandas as pd
import matplotlib.pyplot as plt


df = pd.read_csv("norwegian_financials_raw.csv")

df
```

## 2.1 Profitability Ratio

### Operating Profit Margin



```python
df["operating_profit_margin"] = (
    df["operating_profit_ebit_nok_m"]
    / df["operating_revenue_nok_m"]
) * 100

profit_margin_table = df.loc[
    df["year"].between(2021, 2025),
    [
        "year",
        "operating_revenue_nok_m",
        "operating_profit_ebit_nok_m",
        "operating_profit_margin"
    ]
].copy()

profit_margin_table
```

### Return on Capital Employed (ROCE)

```python
##Calculating Capital Employed
df["capital_employed"] = (
    df["total_equity_nok_m"]
    + df["total_noncurrent_liabilities_nok_m"]
)

df[[
    "year",
    "total_equity_nok_m",
    "total_noncurrent_liabilities_nok_m",
    "capital_employed"
]]

## calculate the average capital employed for each year and the ROCE for each year

df["average_capital_employed"] = (
    df["capital_employed"]
    + df["capital_employed"].shift(1)
) / 2

## ROCE Calculation:
df["roce"] = (
    df["operating_profit_ebit_nok_m"]
    / df["average_capital_employed"]
) * 100

##Displaying the ROCE table for the years 2021-2025

roce_table = df.loc[
    df["year"].between(2021, 2025),
    [
        "year",
        "operating_profit_ebit_nok_m",
        "capital_employed",
        "average_capital_employed",
        "roce"
    ]
].copy()

roce_table
```

```python
## Profitability table (Operating Profit Margin and ROCE) for the years 2021-2025

profitability = df.loc[
    df["year"].between(2021, 2025),
    [
        "year",
        "operating_profit_margin",
        "roce"
    ]
].copy()

profitability = profitability.rename(columns={
    "operating_profit_margin": "Operating Profit Margin (%)",
    "roce": "ROCE (%)"
})

profitability.round(2)
```

## 2.2 Efficiency

### Sales Revenue to Capital Employed


```python
df["sales_to_capital_employed"] = (
    df["operating_revenue_nok_m"]
    / df["average_capital_employed"]
)

sales_capital_table = df.loc[
    df["year"].between(2021, 2025),
    [
        "year",
        "operating_revenue_nok_m",
        "average_capital_employed",
        "sales_to_capital_employed"
    ]
].copy()

sales_capital_table
```

### Sales Revenue per Employee

```python
df["revenue_per_employee_nok_m"] = (
    df["operating_revenue_nok_m"]
    / df["employee_equivalent"]
)

employee_efficiency_table = df.loc[
    df["year"].between(2021, 2025),
    [
        "year",
        "operating_revenue_nok_m",
        "employee_equivalent",
        "employee_measure",
        "revenue_per_employee_nok_m"
    ]
].copy()

employee_efficiency_table
```

```python
##Final Efficiency table (Sales to Capital Employed and Revenue per Employee) for the years 2021-2025

efficiency = df.loc[
    df["year"].between(2021, 2025),
    [
        "year",
        "sales_to_capital_employed",
        "revenue_per_employee_nok_m"
    ]
].copy()

efficiency = efficiency.rename(columns={
    "sales_to_capital_employed": "Sales Revenue to Capital Employed (x)",
    "revenue_per_employee_nok_m": "Revenue per Employee (NOK m)"
})

efficiency.round(2)
```

### 2.3 Liquidity

## Current Ratio

```python
df["current_ratio"] = (
    df["current_assets_nok_m"]
    / df["current_liabilities_nok_m"]
)

current_ratio_table = df.loc[
    df["year"].between(2021, 2025),
    [
        "year",
        "current_assets_nok_m",
        "current_liabilities_nok_m",
        "current_ratio"
    ]
].copy()

current_ratio_table
```

### Acid-Test Ratio

```python
df["acid_test_ratio"] = (
    (df["current_assets_nok_m"] - df["inventory_nok_m"])
    / df["current_liabilities_nok_m"]
)

acid_test_table = df.loc[
    df["year"].between(2021, 2025),
    [
        "year",
        "current_assets_nok_m",
        "inventory_nok_m",
        "current_liabilities_nok_m",
        "acid_test_ratio"
    ]
].copy()

acid_test_table
```

```python
## Final Liquidity table (Current Ratio and Acid Test Ratio) for the years 2021-2025
liquidity = df.loc[
    df["year"].between(2021, 2025),
    [
        "year",
        "current_ratio",
        "acid_test_ratio"
    ]
].copy()

liquidity = liquidity.rename(columns={
    "current_ratio": "Current Ratio (x)",
    "acid_test_ratio": "Acid-Test Ratio (x)"
})

liquidity.round(2)
```

### 2.4 Financial Gearing

### Gearing Ratio

```python
df["noncurrent_interest_bearing_debt"] = (
    df["noncurrent_borrowings_nok_m"]
    + df["noncurrent_lease_liabilities_nok_m"]
)

df["gearing_ratio"] = (
    df["noncurrent_interest_bearing_debt"]
    / (
        df["total_equity_nok_m"]
        + df["noncurrent_interest_bearing_debt"]
    )
) * 100

### Final Gearing table (Gearing Ratio) for the years 2021-2025

gearing_table = df.loc[
    df["year"].between(2021, 2025),
    [
        "year",
        "total_equity_nok_m",
        "noncurrent_borrowings_nok_m",
        "noncurrent_lease_liabilities_nok_m",
        "noncurrent_interest_bearing_debt",
        "gearing_ratio"
    ]
].copy()

gearing_table
```

### Interest Cover Ratio


```python
df["interest_cover"] = (
    df["operating_profit_ebit_nok_m"]
    / df["interest_expense_nok_m"]
)

interest_cover_table = df.loc[
    df["year"].between(2021, 2025),
    [
        "year",
        "operating_profit_ebit_nok_m",
        "interest_expense_nok_m",
        "interest_cover"
    ]
].copy()

interest_cover_table
```

```python
## Financial gearing table (Gearing Ratio and Interest Cover) for the years 2021-2025
financial_gearing = df.loc[
    df["year"].between(2021, 2025),
    [
        "year",
        "gearing_ratio",
        "interest_cover"
    ]
].copy()

financial_gearing = financial_gearing.rename(columns={
    "gearing_ratio": "Gearing Ratio (%)",
    "interest_cover": "Interest Cover (x)"
})

financial_gearing.round(2)
```
